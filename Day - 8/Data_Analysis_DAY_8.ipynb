```python
category_sales = df.groupby("Category")["Total_Sales"].sum()

category_sales.sort_values(ascending=False)
```

```python
city_sales = df.groupby("City")["Total_Sales"].sum()

city_sales.sort_values(ascending=False)
```

```python
product_sales = df.groupby("Product")["Total_Sales"].sum()

product_sales.sort_values(ascending=False)
```

```python
payment_sales = df.groupby("Payment_Method")["Total_Sales"].sum()

payment_sales.sort_values(ascending=False)
```

```python
total_sales = df["Total_Sales"].sum()

print("Total Sales:", total_sales)
```

```python
average_sales = df["Total_Sales"].mean()

print("Average Sales:", average_sales)
```

```python
print("Highest Sale:", df["Total_Sales"].max())
print("Lowest Sale:", df["Total_Sales"].min())
```

```python
total_quantity = df["Quantity"].sum()

print("Total Quantity Sold:", total_quantity)
```

```python
number_of_orders = len(df)

print("Number of Orders:", number_of_orders)
```

```python
df.groupby("Category")["Total_Sales"].agg(
    ["sum", "mean", "max", "min"]
)
```

```python
top_category = (
    df.groupby("Category")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(1)
)

print(top_category)
```

```python
top_product = (
    df.groupby("Product")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(1)
)

print(top_product)
```
