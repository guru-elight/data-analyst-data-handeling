```python
# =====================================================================
# DAY 15 - EXECUTIVE HOTEL BOOKING EDA REPORT
# DATASET-SPECIFIC COMPLETE ONE-CELL SOLUTION
# =====================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import warnings

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

# =====================================================================
# 1. LOAD DATASET
# =====================================================================

FILE_NAME = "Day15_Executive_Hotel_Booking_EDA_Dataset.csv"

if not os.path.exists(FILE_NAME):
    try:
        from google.colab import files
        print("Upload the Day 15 dataset:")
        uploaded = files.upload()
        FILE_NAME = list(uploaded.keys())[0]
    except:
        raise FileNotFoundError(
            "Dataset not found. Place the CSV file in the notebook folder."
        )

df = pd.read_csv(FILE_NAME)

print("=" * 90)
print("        DAY 15 - EXECUTIVE HOTEL BOOKING EDA REPORT")
print("=" * 90)

print("\nDataset loaded successfully!")
print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]}")


# =====================================================================
# 2. INITIAL DATA UNDERSTANDING
# =====================================================================

print("\n" + "=" * 90)
print("1. INITIAL DATA UNDERSTANDING")
print("=" * 90)

print("\nFirst 5 Records:")
display(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
display(pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str).values,
    "Unique_Values": [df[c].nunique() for c in df.columns]
}))

print("\nStatistical Summary:")
display(df.describe(include="all").T)


# =====================================================================
# 3. DATA QUALITY CHECK
# =====================================================================

print("\n" + "=" * 90)
print("2. DATA QUALITY ASSESSMENT")
print("=" * 90)

# Missing values
missing = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

missing = missing[
    missing["Missing_Count"] > 0
].sort_values(
    "Missing_Percentage",
    ascending=False
)

print("\nMissing Values:")
if len(missing) > 0:
    display(missing)
else:
    print("No missing values found.")

# Duplicates
print("\nDuplicate Records:")
print("Exact duplicate rows :", df.duplicated().sum())
print("Duplicate Booking IDs:", df["Booking_ID"].duplicated().sum())

# Unique values in categorical columns
print("\nCategorical Value Inspection:")

categorical_cols = df.select_dtypes(include="object").columns

for col in categorical_cols:
    print(f"\n{col}:")
    print(df[col].value_counts(dropna=False).head(15))


# =====================================================================
# 4. DATA CLEANING
# =====================================================================

print("\n" + "=" * 90)
print("3. DATA CLEANING AND PREPROCESSING")
print("=" * 90)

data = df.copy()

rows_before = len(data)

# -------------------------------------------------
# Remove exact duplicate records
# -------------------------------------------------

data = data.drop_duplicates().reset_index(drop=True)

print(
    f"\nExact duplicate rows removed: "
    f"{rows_before - len(data)}"
)

# -------------------------------------------------
# Clean text columns
# -------------------------------------------------

text_columns = [
    "Hotel_Type",
    "Hotel_Location",
    "Country",
    "Market_Segment",
    "Distribution_Channel",
    "Room_Type_Reserved",
    "Room_Type_Assigned",
    "Deposit_Type",
    "Customer_Type",
    "Meal_Type",
    "Reservation_Status"
]

for col in text_columns:
    if col in data.columns:
        data[col] = (
            data[col]
            .astype("string")
            .str.strip()
        )

# -------------------------------------------------
# Standardize Hotel Type
# -------------------------------------------------

if "Hotel_Type" in data.columns:

    data["Hotel_Type"] = (
        data["Hotel_Type"]
        .str.lower()
        .str.strip()
        .replace({
            "city hotel": "City Hotel",
            "resort hotel": "Resort Hotel"
        })
    )

# -------------------------------------------------
# Standardize Market Segment
# -------------------------------------------------

if "Market_Segment" in data.columns:

    data["Market_Segment"] = (
        data["Market_Segment"]
        .str.lower()
        .str.strip()
        .replace({
            "online ta": "Online TA",
            "offline ta/to": "Offline TA/TO",
            "direct": "Direct",
            "corporate": "Corporate",
            "groups": "Groups"
        })
    )

# -------------------------------------------------
# Standardize Meal Type
# -------------------------------------------------

if "Meal_Type" in data.columns:

    data["Meal_Type"] = (
        data["Meal_Type"]
        .str.lower()
        .str.strip()
        .replace({
            "bb": "BB",
            "b&b": "BB",
            "b & b": "BB",
            "b and b": "BB",
            "hb": "HB",
            "fb": "FB",
            "sc": "SC",
            "undefined": "Undefined",
            "unknown": "Unknown"
        })
    )

# -------------------------------------------------
# Convert dates
# -------------------------------------------------

date_columns = [
    "Booking_Date",
    "Arrival_Date",
    "Reservation_Status_Date"
]

for col in date_columns:
    if col in data.columns:
        data[col] = pd.to_datetime(
            data[col],
            errors="coerce"
        )

# -------------------------------------------------
# Handle missing numeric values
# -------------------------------------------------

numeric_columns = data.select_dtypes(
    include=np.number
).columns

for col in numeric_columns:

    if data[col].isnull().sum() > 0:

        # ID columns -> 0 indicates unavailable
        if col in ["Agent_ID", "Company_ID"]:
            data[col] = data[col].fillna(0)

        else:
            data[col] = data[col].fillna(
                data[col].median()
            )

# -------------------------------------------------
# Handle missing categorical values
# -------------------------------------------------

categorical_columns = data.select_dtypes(
    include=["object", "string"]
).columns

for col in categorical_columns:

    if data[col].isnull().sum() > 0:
        data[col] = data[col].fillna("Unknown")


# =====================================================================
# 5. DATA VALIDATION
# =====================================================================

print("\n" + "=" * 90)
print("4. DATA VALIDATION")
print("=" * 90)

# Total nights validation
if all(
    c in data.columns
    for c in ["Total_Nights", "Weekend_Nights", "Weekday_Nights"]
):

    expected_nights = (
        data["Weekend_Nights"] +
        data["Weekday_Nights"]
    )

    mismatch_nights = (
        data["Total_Nights"] != expected_nights
    ).sum()

    print(
        "Total_Nights inconsistencies:",
        mismatch_nights
    )

# Revenue validation
if all(
    c in data.columns
    for c in ["ADR", "Total_Nights", "Estimated_Revenue"]
):

    calculated_revenue = (
        data["ADR"] *
        data["Total_Nights"]
    )

    revenue_difference = (
        data["Estimated_Revenue"] -
        calculated_revenue
    )

    print(
        "Revenue calculation differences:",
        (abs(revenue_difference) > 0.01).sum()
    )

# Logical checks
print(
    "Negative ADR records:",
    (data["ADR"] < 0).sum()
)

print(
    "Zero-night bookings:",
    (data["Total_Nights"] == 0).sum()
)

print(
    "Invalid guest records:",
    (
        (data["Adults"] <= 0) &
        (data["Children"] <= 0) &
        (data["Babies"] <= 0)
    ).sum()
)


# =====================================================================
# 6. DERIVED BUSINESS FEATURES
# =====================================================================

print("\n" + "=" * 90)
print("5. FEATURE ENGINEERING")
print("=" * 90)

# Total guests
data["Total_Guests"] = (
    data["Adults"] +
    data["Children"] +
    data["Babies"]
)

# Total stay nights
data["Calculated_Nights"] = (
    data["Weekend_Nights"] +
    data["Weekday_Nights"]
)

# Booking year
data["Booking_Year"] = (
    data["Booking_Date"].dt.year
)

# Arrival year
data["Arrival_Year"] = (
    data["Arrival_Date"].dt.year
)

# Arrival month
data["Arrival_Month"] = (
    data["Arrival_Date"].dt.month_name()
)

# Booking month
data["Booking_Month"] = (
    data["Booking_Date"].dt.month_name()
)

# Revenue per guest
data["Revenue_Per_Guest"] = np.where(
    data["Total_Guests"] > 0,
    data["Estimated_Revenue"] /
    data["Total_Guests"],
    0
)

# Cancellation label
data["Booking_Status"] = np.where(
    data["Is_Canceled"] == 1,
    "Canceled",
    "Confirmed"
)

# Lead time category
data["Lead_Time_Category"] = pd.cut(
    data["Lead_Time_Days"],
    bins=[-1, 7, 30, 90, 180, 365, np.inf],
    labels=[
        "0-7 Days",
        "8-30 Days",
        "31-90 Days",
        "91-180 Days",
        "181-365 Days",
        "365+ Days"
    ]
)

print("Business features created successfully.")


# =====================================================================
# 7. OUTLIER ANALYSIS
# =====================================================================

print("\n" + "=" * 90)
print("6. OUTLIER ANALYSIS")
print("=" * 90)

outlier_cols = [
    "Lead_Time_Days",
    "ADR",
    "Total_Nights",
    "Estimated_Revenue",
    "Satisfaction_Score",
    "Days_In_Waiting_List"
]

outlier_results = []

for col in outlier_cols:

    if col in data.columns:

        Q1 = data[col].quantile(0.25)
        Q3 = data[col].quantile(0.75)

        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        count = (
            (data[col] < lower) |
            (data[col] > upper)
        ).sum()

        outlier_results.append([
            col,
            round(Q1, 2),
            round(Q3, 2),
            round(lower, 2),
            round(upper, 2),
            count,
            round(count / len(data) * 100, 2)
        ])

outlier_df = pd.DataFrame(
    outlier_results,
    columns=[
        "Variable",
        "Q1",
        "Q3",
        "Lower_Bound",
        "Upper_Bound",
        "Outlier_Count",
        "Outlier_Percentage"
    ]
)

display(outlier_df)


# =====================================================================
# 8. EXECUTIVE KPI DASHBOARD
# =====================================================================

print("\n" + "=" * 90)
print("7. EXECUTIVE KPI DASHBOARD")
print("=" * 90)

total_bookings = len(data)

cancellation_rate = (
    data["Is_Canceled"].mean() * 100
)

avg_adr = data["ADR"].mean()

median_adr = data["ADR"].median()

avg_lead_time = data["Lead_Time_Days"].mean()

avg_stay = data["Total_Nights"].mean()

avg_guests = data["Total_Guests"].mean()

total_revenue = data["Estimated_Revenue"].sum()

avg_revenue = data["Estimated_Revenue"].mean()

avg_satisfaction = data["Satisfaction_Score"].mean()

repeat_guest_rate = (
    data["Is_Repeated_Guest"].mean() * 100
)

print(f"""
Total Bookings              : {total_bookings:,}
Cancellation Rate           : {cancellation_rate:.2f}%
Average ADR                 : {avg_adr:.2f}
Median ADR                  : {median_adr:.2f}
Average Lead Time           : {avg_lead_time:.2f} days
Average Stay                : {avg_stay:.2f} nights
Average Guests              : {avg_guests:.2f}
Total Estimated Revenue     : {total_revenue:,.2f}
Average Revenue / Booking   : {avg_revenue:,.2f}
Average Satisfaction Score  : {avg_satisfaction:.2f}
Repeat Guest Rate           : {repeat_guest_rate:.2f}%
""")


# =====================================================================
# 9. UNIVARIATE ANALYSIS
# =====================================================================

print("\n" + "=" * 90)
print("8. UNIVARIATE ANALYSIS")
print("=" * 90)

# -------------------------------------------------
# Hotel type
# -------------------------------------------------

plt.figure(figsize=(8, 5))

sns.countplot(
    data=data,
    x="Hotel_Type",
    order=data["Hotel_Type"].value_counts().index
)

plt.title("Bookings by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Number of Bookings")
plt.xticks(rotation=10)
plt.tight_layout()
plt.show()


# -------------------------------------------------
# Booking status
# -------------------------------------------------

plt.figure(figsize=(8, 5))

sns.countplot(
    data=data,
    x="Booking_Status",
    order=data["Booking_Status"].value_counts().index
)

plt.title("Booking Cancellation Status")
plt.xlabel("Booking Status")
plt.ylabel("Number of Bookings")
plt.tight_layout()
plt.show()


# -------------------------------------------------
# Market segment
# -------------------------------------------------

plt.figure(figsize=(10, 5))

segment_counts = (
    data["Market_Segment"]
    .value_counts()
    .head(10)
)

sns.barplot(
    x=segment_counts.values,
    y=segment_counts.index
)

plt.title("Bookings by Market Segment")
plt.xlabel("Number of Bookings")
plt.ylabel("Market Segment")
plt.tight_layout()
plt.show()


# -------------------------------------------------
# ADR distribution
# -------------------------------------------------

plt.figure(figsize=(10, 5))

sns.histplot(
    data["ADR"],
    bins=40,
    kde=True
)

plt.title("Distribution of Average Daily Rate (ADR)")
plt.xlabel("ADR")
plt.ylabel("Frequency")

plt.xlim(
    0,
    data["ADR"].quantile(0.99)
)

plt.tight_layout()
plt.show()


# -------------------------------------------------
# Lead time distribution
# -------------------------------------------------

plt.figure(figsize=(10, 5))

sns.histplot(
    data["Lead_Time_Days"],
    bins=40,
    kde=True
)

plt.title("Distribution of Lead Time")
plt.xlabel("Lead Time (Days)")
plt.ylabel("Frequency")

plt.xlim(
    0,
    data["Lead_Time_Days"].quantile(0.99)
)

plt.tight_layout()
plt.show()


# =====================================================================
# 10. BIVARIATE ANALYSIS
# =====================================================================

print("\n" + "=" * 90)
print("9. BIVARIATE ANALYSIS")
print("=" * 90)


# -------------------------------------------------
# Cancellation by hotel
# -------------------------------------------------

hotel_cancel = (
    data.groupby("Hotel_Type")["Is_Canceled"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

print("\nCancellation Rate by Hotel Type:")
display(
    hotel_cancel.to_frame(
        "Cancellation_Rate_Percentage"
    ).round(2)
)

plt.figure(figsize=(8, 5))

sns.barplot(
    x=hotel_cancel.index,
    y=hotel_cancel.values
)

plt.title("Cancellation Rate by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Cancellation Rate (%)")
plt.tight_layout()
plt.show()


# -------------------------------------------------
# Cancellation by market segment
# -------------------------------------------------

segment_cancel = (
    data.groupby("Market_Segment")["Is_Canceled"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

print("\nCancellation Rate by Market Segment:")
display(
    segment_cancel.to_frame(
        "Cancellation_Rate_Percentage"
    ).round(2)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=segment_cancel.values,
    y=segment_cancel.index
)

plt.title("Cancellation Rate by Market Segment")
plt.xlabel("Cancellation Rate (%)")
plt.ylabel("Market Segment")
plt.tight_layout()
plt.show()


# -------------------------------------------------
# Cancellation by deposit type
# -------------------------------------------------

deposit_cancel = (
    data.groupby("Deposit_Type")["Is_Canceled"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

print("\nCancellation Rate by Deposit Type:")
display(
    deposit_cancel.to_frame(
        "Cancellation_Rate_Percentage"
    ).round(2)
)

plt.figure(figsize=(9, 5))

sns.barplot(
    x=deposit_cancel.index,
    y=deposit_cancel.values
)

plt.title("Cancellation Rate by Deposit Type")
plt.xlabel("Deposit Type")
plt.ylabel("Cancellation Rate (%)")
plt.tight_layout()
plt.show()


# -------------------------------------------------
# ADR by hotel
# -------------------------------------------------

adr_hotel = (
    data.groupby("Hotel_Type")["ADR"]
    .agg(["mean", "median", "min", "max"])
    .round(2)
)

print("\nADR Statistics by Hotel Type:")
display(adr_hotel)

plt.figure(figsize=(9, 5))

sns.boxplot(
    data=data,
    x="Hotel_Type",
    y="ADR"
)

plt.title("ADR Distribution by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("ADR")

plt.ylim(
    0,
    data["ADR"].quantile(0.99)
)

plt.tight_layout()
plt.show()


# =====================================================================
# 11. LEAD TIME VS CANCELLATION
# =====================================================================

lead_cancel = (
    data.groupby("Lead_Time_Category", observed=False)
    .agg(
        Bookings=("Booking_ID", "count"),
        Cancellation_Rate=("Is_Canceled", "mean")
    )
)

lead_cancel["Cancellation_Rate"] *= 100

print("\nCancellation Rate by Lead-Time Category:")
display(lead_cancel.round(2))

plt.figure(figsize=(11, 5))

sns.barplot(
    data=lead_cancel.reset_index(),
    x="Lead_Time_Category",
    y="Cancellation_Rate"
)

plt.title("Lead Time vs Cancellation Rate")
plt.xlabel("Lead Time Category")
plt.ylabel("Cancellation Rate (%)")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()


# =====================================================================
# 12. CUSTOMER TYPE ANALYSIS
# =====================================================================

customer_analysis = (
    data.groupby("Customer_Type")
    .agg(
        Bookings=("Booking_ID", "count"),
        Cancellation_Rate=("Is_Canceled", "mean"),
        Average_ADR=("ADR", "mean"),
        Average_Stay=("Total_Nights", "mean"),
        Average_Satisfaction=("Satisfaction_Score", "mean")
    )
)

customer_analysis["Cancellation_Rate"] *= 100

print("\nCustomer Type Performance:")
display(customer_analysis.round(2))

plt.figure(figsize=(10, 5))

sns.barplot(
    data=customer_analysis.reset_index(),
    x="Customer_Type",
    y="Cancellation_Rate"
)

plt.title("Cancellation Rate by Customer Type")
plt.xlabel("Customer Type")
plt.ylabel("Cancellation Rate (%)")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()


# =====================================================================
# 13. MONTHLY ANALYSIS
# =====================================================================

month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

data["Arrival_Month"] = pd.Categorical(
    data["Arrival_Month"],
    categories=month_order,
    ordered=True
)

monthly_analysis = (
    data.groupby("Arrival_Month", observed=False)
    .agg(
        Bookings=("Booking_ID", "count"),
        Average_ADR=("ADR", "mean"),
        Cancellation_Rate=("Is_Canceled", "mean"),
        Estimated_Revenue=("Estimated_Revenue", "sum")
    )
)

monthly_analysis["Cancellation_Rate"] *= 100

print("\nMonthly Performance:")
display(monthly_analysis.round(2))


# Monthly booking volume
plt.figure(figsize=(13, 5))

sns.barplot(
    x=monthly_analysis.index,
    y=monthly_analysis["Bookings"]
)

plt.title("Monthly Booking Volume")
plt.xlabel("Arrival Month")
plt.ylabel("Bookings")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# Monthly ADR
plt.figure(figsize=(13, 5))

sns.lineplot(
    x=monthly_analysis.index,
    y=monthly_analysis["Average_ADR"],
    marker="o"
)

plt.title("Average ADR by Month")
plt.xlabel("Arrival Month")
plt.ylabel("Average ADR")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# Monthly cancellation
plt.figure(figsize=(13, 5))

sns.lineplot(
    x=monthly_analysis.index,
    y=monthly_analysis["Cancellation_Rate"],
    marker="o"
)

plt.title("Monthly Cancellation Rate")
plt.xlabel("Arrival Month")
plt.ylabel("Cancellation Rate (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# =====================================================================
# 14. HOTEL × MONTH HEATMAP
# =====================================================================

hotel_month = pd.crosstab(
    data["Hotel_Type"],
    data["Arrival_Month"]
)

hotel_month = hotel_month.reindex(
    columns=month_order
)

plt.figure(figsize=(15, 5))

sns.heatmap(
    hotel_month,
    annot=True,
    fmt="d",
    cmap="YlGnBu"
)

plt.title("Booking Volume by Hotel Type and Month")
plt.xlabel("Arrival Month")
plt.ylabel("Hotel Type")
plt.tight_layout()
plt.show()


# =====================================================================
# 15. ROOM TYPE ANALYSIS
# =====================================================================

room_analysis = (
    data.groupby("Room_Type_Reserved")
    .agg(
        Bookings=("Booking_ID", "count"),
        Average_ADR=("ADR", "mean"),
        Cancellation_Rate=("Is_Canceled", "mean"),
        Average_Stay=("Total_Nights", "mean")
    )
)

room_analysis["Cancellation_Rate"] *= 100

room_analysis = room_analysis.sort_values(
    "Bookings",
    ascending=False
)

print("\nRoom Type Performance:")
display(room_analysis.round(2))

plt.figure(figsize=(10, 5))

sns.barplot(
    data=room_analysis.reset_index(),
    x="Room_Type_Reserved",
    y="Bookings"
)

plt.title("Bookings by Reserved Room Type")
plt.xlabel("Room Type")
plt.ylabel("Bookings")
plt.tight_layout()
plt.show()


# =====================================================================
# 16. REPEAT GUEST ANALYSIS
# =====================================================================

repeat_analysis = (
    data.groupby("Is_Repeated_Guest")
    .agg(
        Bookings=("Booking_ID", "count"),
        Cancellation_Rate=("Is_Canceled", "mean"),
        Average_ADR=("ADR", "mean"),
        Average_Satisfaction=("Satisfaction_Score", "mean"),
        Average_Revenue=("Estimated_Revenue", "mean")
    )
)

repeat_analysis["Cancellation_Rate"] *= 100

repeat_analysis.index = [
    "New Guest" if x == 0 else "Repeated Guest"
    for x in repeat_analysis.index
]

print("\nNew vs Repeated Guest Analysis:")
display(repeat_analysis.round(2))

plt.figure(figsize=(8, 5))

sns.barplot(
    x=repeat_analysis.index,
    y=repeat_analysis["Cancellation_Rate"]
)

plt.title("Cancellation Rate: New vs Repeated Guests")
plt.xlabel("Guest Type")
plt.ylabel("Cancellation Rate (%)")
plt.tight_layout()
plt.show()


# =====================================================================
# 17. SPECIAL REQUEST ANALYSIS
# =====================================================================

request_analysis = (
    data.groupby("Total_Special_Requests")
    .agg(
        Bookings=("Booking_ID", "count"),
        Cancellation_Rate=("Is_Canceled", "mean"),
        Average_Satisfaction=("Satisfaction_Score", "mean")
    )
)

request_analysis["Cancellation_Rate"] *= 100

print("\nSpecial Requests Analysis:")
display(request_analysis.round(2))

plt.figure(figsize=(10, 5))

sns.barplot(
    x=request_analysis.index,
    y=request_analysis["Cancellation_Rate"]
)

plt.title("Special Requests vs Cancellation Rate")
plt.xlabel("Number of Special Requests")
plt.ylabel("Cancellation Rate (%)")
plt.tight_layout()
plt.show()


# =====================================================================
# 18. REVENUE ANALYSIS
# =====================================================================

hotel_revenue = (
    data.groupby("Hotel_Type")
    .agg(
        Bookings=("Booking_ID", "count"),
        Total_Revenue=("Estimated_Revenue", "sum"),
        Average_Revenue=("Estimated_Revenue", "mean"),
        Average_ADR=("ADR", "mean")
    )
    .sort_values(
        "Total_Revenue",
        ascending=False
    )
)

print("\nRevenue Performance by Hotel Type:")
display(hotel_revenue.round(2))

plt.figure(figsize=(9, 5))

sns.barplot(
    data=hotel_revenue.reset_index(),
    x="Hotel_Type",
    y="Total_Revenue"
)

plt.title("Estimated Revenue by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Estimated Revenue")
plt.tight_layout()
plt.show()


# =====================================================================
# 19. TOP COUNTRIES
# =====================================================================

country_analysis = (
    data["Country"]
    .value_counts()
    .head(15)
)

print("\nTop 15 Customer Countries:")
display(
    country_analysis.to_frame("Bookings")
)

plt.figure(figsize=(10, 7))

sns.barplot(
    x=country_analysis.values,
    y=country_analysis.index
)

plt.title("Top 15 Customer Countries")
plt.xlabel("Bookings")
plt.ylabel("Country")
plt.tight_layout()
plt.show()


# =====================================================================
# 20. CORRELATION ANALYSIS
# =====================================================================

print("\n" + "=" * 90)
print("10. CORRELATION ANALYSIS")
print("=" * 90)

correlation_columns = [
    "Lead_Time_Days",
    "Weekend_Nights",
    "Weekday_Nights",
    "Adults",
    "Children",
    "Babies",
    "Is_Repeated_Guest",
    "Previous_Cancellations",
    "Previous_Bookings",
    "Booking_Changes",
    "Days_In_Waiting_List",
    "Total_Nights",
    "ADR",
    "Required_Car_Parking_Spaces",
    "Total_Special_Requests",
    "Satisfaction_Score",
    "Is_Canceled",
    "Estimated_Revenue",
    "Total_Guests"
]

correlation_columns = [
    c for c in correlation_columns
    if c in data.columns
]

corr = data[correlation_columns].corr()

plt.figure(figsize=(16, 12))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)

plt.title("Correlation Heatmap - Hotel Booking Dataset")
plt.tight_layout()
plt.show()


# Strong correlations
corr_pairs = (
    corr.where(
        np.triu(
            np.ones(corr.shape),
            k=1
        ).astype(bool)
    )
    .stack()
    .sort_values(
        key=lambda x: abs(x),
        ascending=False
    )
)

print("\nTop 15 Correlations:")
display(
    corr_pairs.head(15).to_frame(
        "Correlation"
    )
)


# =====================================================================
# 21. LEAD TIME VS ADR
# =====================================================================

plt.figure(figsize=(10, 6))

sample_data = data.sample(
    min(10000, len(data)),
    random_state=42
)

sns.scatterplot(
    data=sample_data,
    x="Lead_Time_Days",
    y="ADR",
    hue="Hotel_Type",
    alpha=0.5
)

plt.title("Lead Time vs ADR")
plt.xlabel("Lead Time (Days)")
plt.ylabel("ADR")

plt.xlim(
    0,
    data["Lead_Time_Days"].quantile(0.99)
)

plt.ylim(
    0,
    data["ADR"].quantile(0.99)
)

plt.tight_layout()
plt.show()


# =====================================================================
# 22. CANCELLATION BY REVENUE / ADR
# =====================================================================

plt.figure(figsize=(9, 6))

sns.boxplot(
    data=data,
    x="Booking_Status",
    y="ADR"
)

plt.title("ADR Distribution by Booking Status")
plt.xlabel("Booking Status")
plt.ylabel("ADR")

plt.ylim(
    0,
    data["ADR"].quantile(0.99)
)

plt.tight_layout()
plt.show()


# =====================================================================
# 23. DATASET QUALITY AFTER CLEANING
# =====================================================================

print("\n" + "=" * 90)
print("11. FINAL DATA QUALITY CHECK")
print("=" * 90)

print("Rows after cleaning       :", len(data))
print("Columns after engineering :", len(data.columns))
print("Remaining missing values  :", data.isnull().sum().sum())
print("Remaining duplicate rows  :", data.duplicated().sum())

print("\nMissing Values After Cleaning:")
remaining_missing = data.isnull().sum()

remaining_missing = remaining_missing[
    remaining_missing > 0
]

if len(remaining_missing) > 0:
    display(remaining_missing)
else:
    print("No missing values remain.")


# =====================================================================
# 24. AUTOMATIC BUSINESS INSIGHTS
# =====================================================================

print("\n" + "=" * 90)
print("12. KEY BUSINESS INSIGHTS")
print("=" * 90)

# Insight 1
highest_hotel_cancel = hotel_cancel.idxmax()
highest_hotel_cancel_rate = hotel_cancel.max()

# Insight 2
highest_segment_cancel = segment_cancel.idxmax()
highest_segment_cancel_rate = segment_cancel.max()

# Insight 3
highest_month = monthly_analysis[
    "Bookings"
].idxmax()

highest_month_bookings = monthly_analysis[
    "Bookings"
].max()

# Insight 4
highest_revenue_hotel = hotel_revenue[
    "Total_Revenue"
].idxmax()

# Insight 5
highest_lead_category = lead_cancel[
    "Cancellation_Rate"
].idxmax()

highest_lead_cancel_rate = lead_cancel[
    "Cancellation_Rate"
].max()

# Insight 6
highest_adr_hotel = (
    hotel_revenue["Average_ADR"]
    .idxmax()
)

print(f"""
INSIGHT 1:
The overall cancellation rate is {cancellation_rate:.2f}%.
This indicates that cancellation management is a major revenue-protection
opportunity for hotel management.

INSIGHT 2:
{highest_hotel_cancel} has the highest hotel-level cancellation rate
at approximately {highest_hotel_cancel_rate:.2f}%.
Hotel-specific cancellation policies should therefore be considered.

INSIGHT 3:
The {highest_segment_cancel} market segment has the highest cancellation
rate at approximately {highest_segment_cancel_rate:.2f}%.
This segment should receive additional monitoring and targeted booking policies.

INSIGHT 4:
{highest_month} records the highest booking volume with approximately
{highest_month_bookings:,} bookings, highlighting an important demand period
for inventory and staffing planning.

INSIGHT 5:
Bookings in the {highest_lead_category} lead-time category show the highest
cancellation tendency at approximately {highest_lead_cancel_rate:.2f}%.
Longer booking horizons should therefore be monitored carefully.

INSIGHT 6:
{highest_revenue_hotel} contributes the highest estimated revenue among
the hotel types in the dataset.

INSIGHT 7:
The average ADR is approximately {avg_adr:.2f}, while the average stay is
{avg_stay:.2f} nights. Pricing and length-of-stay strategies can therefore
be combined to improve revenue per booking.
""")


# =====================================================================
# 25. MANAGEMENT RECOMMENDATIONS
# =====================================================================

print("\n" + "=" * 90)
print("13. MANAGEMENT RECOMMENDATIONS")
print("=" * 90)

recommendations = [
    "1. Introduce risk-based cancellation policies for bookings with high "
    "lead times and historically high cancellation probability.",

    "2. Use deposits or partial prepayments selectively for high-risk "
    "segments rather than applying strict policies to every customer.",

    "3. Implement dynamic pricing based on monthly demand, ADR patterns, "
    "hotel type, and booking lead time.",

    "4. Strengthen customer retention programs for repeat guests because "
    "repeat customers can provide more predictable demand and revenue.",

    "5. Allocate rooms, staff, and operational resources according to "
    "seasonal booking patterns and peak-demand months.",

    "6. Monitor market segments with high cancellation rates and create "
    "segment-specific offers, policies, and communication strategies.",

    "7. Use booking, cancellation, ADR, customer, and revenue KPIs in a "
    "regular management dashboard to support data-driven decisions."
]

for recommendation in recommendations:
    print("\n" + recommendation)


# =====================================================================
# 26. FINAL EXECUTIVE TABLE
# =====================================================================

print("\n" + "=" * 90)
print("14. EXECUTIVE SUMMARY TABLE")
print("=" * 90)

executive_summary = pd.DataFrame({
    "KPI": [
        "Total Bookings",
        "Cancellation Rate",
        "Average ADR",
        "Average Lead Time",
        "Average Stay",
        "Average Guests",
        "Total Estimated Revenue",
        "Average Revenue per Booking",
        "Average Satisfaction",
        "Repeat Guest Rate"
    ],
    "Value": [
        f"{total_bookings:,}",
        f"{cancellation_rate:.2f}%",
        f"{avg_adr:.2f}",
        f"{avg_lead_time:.2f} days",
        f"{avg_stay:.2f} nights",
        f"{avg_guests:.2f}",
        f"{total_revenue:,.2f}",
        f"{avg_revenue:,.2f}",
        f"{avg_satisfaction:.2f}",
        f"{repeat_guest_rate:.2f}%"
    ]
})

display(executive_summary)


# =====================================================================
# 27. SAVE CLEANED DATASET
# =====================================================================

OUTPUT_FILE = "Day15_Cleaned_Hotel_Booking_EDA.csv"

data.to_csv(
    OUTPUT_FILE,
    index=False
)

print("\nCleaned dataset saved as:")
print(OUTPUT_FILE)


# =====================================================================
# 28. FINAL REPORT
# =====================================================================

print("\n" + "=" * 90)
print("                    EDA COMPLETED SUCCESSFULLY")
print("=" * 90)

print(f"""
FINAL DATASET
-------------
Original Records       : {len(df):,}
Cleaned Records        : {len(data):,}
Original Columns       : {df.shape[1]}
Final Columns          : {data.shape[1]}
Duplicates Removed     : {df.duplicated().sum()}
Missing Values Handled : Yes

ANALYSIS COMPLETED
------------------
✓ Data Understanding
✓ Data Quality Assessment
✓ Missing Value Analysis
✓ Duplicate Detection
✓ Inconsistent Value Detection
✓ Data Type Conversion
✓ Text Standardization
✓ Date Processing
✓ Feature Engineering
✓ Outlier Detection
✓ Descriptive Statistics
✓ Univariate Analysis
✓ Bivariate Analysis
✓ Group-wise Analysis
✓ Seasonal Analysis
✓ Revenue Analysis
✓ Customer Analysis
✓ Correlation Analysis
✓ Business Insights
✓ Management Recommendations
✓ Executive KPI Summary
✓ Cleaned Dataset Export

This notebook is ready to be submitted as the Day 15
Executive Hotel Booking EDA Report.
""")

print("=" * 90)
```
