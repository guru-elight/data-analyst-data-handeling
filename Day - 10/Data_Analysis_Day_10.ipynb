```python
# ============================================================
# ASSIGNMENT: FLIGHT OPERATIONS DATA ANALYSIS - DAY 10
# Complete Pandas Analysis - Single Cell
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# 1. LOAD DATASET
# ============================================================

file_path = "/content/Day10_Flight_Operations_Dataset.csv"

# If running outside Google Colab, use the uploaded/local path
import os
if not os.path.exists(file_path):
    file_path = "/mnt/data/Day10_Flight_Operations_Dataset.csv"

df = pd.read_csv(file_path)

print("=" * 80)
print("FLIGHT OPERATIONS DATA ANALYSIS - DAY 10")
print("=" * 80)

print("\nDataset loaded successfully!")
print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")


# ============================================================
# 2. BASIC DATASET UNDERSTANDING
# ============================================================

print("\n" + "=" * 80)
print("1. DATASET OVERVIEW")
print("=" * 80)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Records:")
display(df.head())

print("\nLast 5 Records:")
display(df.tail())

print("\nDataset Shape:")
print(df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()


# ============================================================
# 3. CHECK MISSING VALUES AND DUPLICATES
# ============================================================

print("\n" + "=" * 80)
print("2. DATA QUALITY CHECK")
print("=" * 80)

print("\nMissing Values:")
missing_values = df.isnull().sum()
print(missing_values)

print("\nTotal Missing Values:", df.isnull().sum().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nUnique Values Per Column:")
print(df.nunique())


# ============================================================
# 4. HANDLE MISSING VALUES
# ============================================================

# Delay_Minutes has missing values.
# Since 0 minutes represents no delay, missing delay values
# are replaced using the median delay.

if df["Delay_Minutes"].isnull().sum() > 0:
    median_delay = df["Delay_Minutes"].median()
    df["Delay_Minutes"] = df["Delay_Minutes"].fillna(median_delay)

print("\nMissing values handled successfully.")
print("Remaining missing values:", df.isnull().sum().sum())


# ============================================================
# 5. DATA TYPE CONVERSION
# ============================================================

df["Flight_Date"] = pd.to_datetime(df["Flight_Date"])

print("\nFlight_Date converted to datetime.")

print("\nUpdated Data Types:")
print(df.dtypes)


# ============================================================
# 6. DESCRIPTIVE STATISTICS
# ============================================================

print("\n" + "=" * 80)
print("3. DESCRIPTIVE STATISTICS")
print("=" * 80)

print("\nNumerical Summary:")
display(df.describe())

print("\nCategorical Summary:")
display(df.describe(include="object"))


# ============================================================
# 7. SELECTING RELEVANT DATA
# ============================================================

print("\n" + "=" * 80)
print("4. DATA SELECTION")
print("=" * 80)

print("\nSelected Columns:")
selected_columns = df[
    [
        "Flight_ID",
        "Airline",
        "Origin",
        "Destination",
        "Passengers",
        "Delay_Minutes",
        "Flight_Status",
        "Passenger_Satisfaction"
    ]
]

display(selected_columns.head(10))


# ============================================================
# 8. FILTERING DATA
# ============================================================

print("\n" + "=" * 80)
print("5. DATA FILTERING")
print("=" * 80)

# Delayed flights
delayed_flights = df[df["Flight_Status"] == "Delayed"]

print("\nDelayed Flights:")
display(
    delayed_flights[
        [
            "Flight_ID",
            "Airline",
            "Origin",
            "Destination",
            "Delay_Minutes",
            "Flight_Status"
        ]
    ].head(10)
)

# Flights with delay greater than 30 minutes
high_delay_flights = df[df["Delay_Minutes"] > 30]

print("\nFlights Delayed More Than 30 Minutes:")
display(
    high_delay_flights[
        [
            "Flight_ID",
            "Airline",
            "Delay_Minutes",
            "Weather",
            "Flight_Status"
        ]
    ]
)

# Highly satisfied passengers
high_satisfaction = df[df["Passenger_Satisfaction"] >= 4]

print("\nFlights With Passenger Satisfaction >= 4:")
display(
    high_satisfaction[
        [
            "Flight_ID",
            "Airline",
            "Passenger_Satisfaction",
            "Travel_Class"
        ]
    ].head(10)
)


# ============================================================
# 9. SORTING
# ============================================================

print("\n" + "=" * 80)
print("6. SORTING")
print("=" * 80)

print("\nTop 10 Flights by Passenger Count:")
top_passenger_flights = df.sort_values(
    by="Passengers",
    ascending=False
)

display(
    top_passenger_flights[
        [
            "Flight_ID",
            "Airline",
            "Passengers",
            "Seat_Capacity"
        ]
    ].head(10)
)

print("\nTop 10 Flights by Delay:")
top_delayed_flights = df.sort_values(
    by="Delay_Minutes",
    ascending=False
)

display(
    top_delayed_flights[
        [
            "Flight_ID",
            "Airline",
            "Delay_Minutes",
            "Weather",
            "Flight_Status"
        ]
    ].head(10)
)


# ============================================================
# 10. GROUPING AND AGGREGATION
# ============================================================

print("\n" + "=" * 80)
print("7. GROUPING AND AGGREGATION")
print("=" * 80)

# Airline performance
airline_analysis = df.groupby("Airline").agg(
    Total_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Passengers=("Passengers", "mean"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Ticket_Price=("Average_Ticket_Price", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values(
    by="Total_Passengers",
    ascending=False
)

print("\nAirline Performance:")
display(airline_analysis.round(2))


# ============================================================
# 11. FLIGHT STATUS ANALYSIS
# ============================================================

status_analysis = df.groupby("Flight_Status").agg(
    Number_of_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values(
    by="Number_of_Flights",
    ascending=False
)

print("\nFlight Status Analysis:")
display(status_analysis.round(2))


# ============================================================
# 12. TRAVEL CLASS ANALYSIS
# ============================================================

class_analysis = df.groupby("Travel_Class").agg(
    Number_of_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Ticket_Price=("Average_Ticket_Price", "mean"),
    Average_Baggage=("Avg_Baggage_Kg", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values(
    by="Total_Passengers",
    ascending=False
)

print("\nTravel Class Analysis:")
display(class_analysis.round(2))


# ============================================================
# 13. WEATHER ANALYSIS
# ============================================================

weather_analysis = df.groupby("Weather").agg(
    Number_of_Flights=("Flight_ID", "count"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values(
    by="Average_Delay",
    ascending=False
)

print("\nWeather Impact Analysis:")
display(weather_analysis.round(2))


# ============================================================
# 14. BOOKING CHANNEL ANALYSIS
# ============================================================

booking_analysis = df.groupby("Booking_Channel").agg(
    Number_of_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Ticket_Price=("Average_Ticket_Price", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values(
    by="Number_of_Flights",
    ascending=False
)

print("\nBooking Channel Analysis:")
display(booking_analysis.round(2))


# ============================================================
# 15. AIRCRAFT ANALYSIS
# ============================================================

aircraft_analysis = df.groupby("Aircraft").agg(
    Number_of_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values(
    by="Total_Passengers",
    ascending=False
)

print("\nAircraft Performance:")
display(aircraft_analysis.round(2))


# ============================================================
# 16. ROUTE ANALYSIS
# ============================================================

route_analysis = df.groupby(
    ["Origin", "Destination"]
).agg(
    Number_of_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values(
    by="Number_of_Flights",
    ascending=False
)

print("\nTop Flight Routes:")
display(route_analysis.head(15).round(2))


# ============================================================
# 17. DATA TRANSFORMATIONS
# ============================================================

print("\n" + "=" * 80)
print("8. DATA TRANSFORMATIONS")
print("=" * 80)

# Delay category
df["Delay_Category"] = pd.cut(
    df["Delay_Minutes"],
    bins=[-1, 0, 15, 30, np.inf],
    labels=["No Delay", "Short Delay", "Moderate Delay", "Long Delay"]
)

# Passenger load factor
df["Load_Factor"] = (
    df["Passengers"] / df["Seat_Capacity"]
) * 100

# Estimated revenue
df["Estimated_Revenue"] = (
    df["Passengers"] * df["Average_Ticket_Price"]
)

# Flight month
df["Flight_Month"] = df["Flight_Date"].dt.month

# Flight day
df["Flight_Day"] = df["Flight_Date"].dt.day

# Flight weekday
df["Weekday"] = df["Flight_Date"].dt.day_name()

print("\nNew Columns Created:")
print([
    "Delay_Category",
    "Load_Factor",
    "Estimated_Revenue",
    "Flight_Month",
    "Flight_Day",
    "Weekday"
])

print("\nTransformed Dataset:")
display(df.head())


# ============================================================
# 18. DELAY CATEGORY ANALYSIS
# ============================================================

delay_category_analysis = df.groupby(
    "Delay_Category",
    observed=True
).agg(
    Number_of_Flights=("Flight_ID", "count"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
)

print("\nDelay Category Analysis:")
display(delay_category_analysis.round(2))


# ============================================================
# 19. MONTHLY ANALYSIS
# ============================================================

monthly_analysis = df.groupby("Flight_Month").agg(
    Number_of_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean"),
    Estimated_Revenue=("Estimated_Revenue", "sum")
)

print("\nMonthly Flight Analysis:")
display(monthly_analysis.round(2))


# ============================================================
# 20. ANSWERING MEANINGFUL QUESTIONS
# ============================================================

print("\n" + "=" * 80)
print("9. KEY QUESTIONS AND ANSWERS")
print("=" * 80)

# Q1
most_used_airline = df["Airline"].value_counts().idxmax()
most_used_airline_count = df["Airline"].value_counts().max()

print(
    f"\nQ1. Which airline operates the most flights?"
    f"\nAnswer: {most_used_airline} with {most_used_airline_count} flights."
)

# Q2
most_common_origin = df["Origin"].value_counts().idxmax()
origin_count = df["Origin"].value_counts().max()

print(
    f"\nQ2. Which origin airport/city appears most frequently?"
    f"\nAnswer: {most_common_origin} with {origin_count} flights."
)

# Q3
most_common_destination = df["Destination"].value_counts().idxmax()
destination_count = df["Destination"].value_counts().max()

print(
    f"\nQ3. Which destination appears most frequently?"
    f"\nAnswer: {most_common_destination} with {destination_count} flights."
)

# Q4
highest_delay_airline = airline_analysis["Average_Delay"].idxmax()
highest_delay_value = airline_analysis["Average_Delay"].max()

print(
    f"\nQ4. Which airline has the highest average delay?"
    f"\nAnswer: {highest_delay_airline} with an average delay of "
    f"{highest_delay_value:.2f} minutes."
)

# Q5
best_satisfaction_airline = airline_analysis["Average_Satisfaction"].idxmax()
best_satisfaction_value = airline_analysis["Average_Satisfaction"].max()

print(
    f"\nQ5. Which airline has the highest average passenger satisfaction?"
    f"\nAnswer: {best_satisfaction_airline} with a satisfaction score of "
    f"{best_satisfaction_value:.2f}/5."
)

# Q6
worst_weather = weather_analysis["Average_Delay"].idxmax()
worst_weather_delay = weather_analysis["Average_Delay"].max()

print(
    f"\nQ6. Which weather condition is associated with the highest average delay?"
    f"\nAnswer: {worst_weather} with {worst_weather_delay:.2f} minutes."
)

# Q7
highest_revenue_airline = df.groupby(
    "Airline"
)["Estimated_Revenue"].sum().idxmax()

highest_revenue = df.groupby(
    "Airline"
)["Estimated_Revenue"].sum().max()

print(
    f"\nQ7. Which airline generates the highest estimated revenue?"
    f"\nAnswer: {highest_revenue_airline} with estimated revenue of "
    f"₹{highest_revenue:,.2f}."
)

# Q8
highest_load_airline = df.groupby(
    "Airline"
)["Load_Factor"].mean().idxmax()

highest_load = df.groupby(
    "Airline"
)["Load_Factor"].mean().max()

print(
    f"\nQ8. Which airline has the highest average passenger load factor?"
    f"\nAnswer: {highest_load_airline} with {highest_load:.2f}%."
)


# ============================================================
# 21. VISUAL ANALYSIS
# ============================================================

print("\n" + "=" * 80)
print("10. VISUAL ANALYSIS")
print("=" * 80)

# Airline flight count
plt.figure(figsize=(10, 5))
df["Airline"].value_counts().plot(kind="bar")
plt.title("Number of Flights by Airline")
plt.xlabel("Airline")
plt.ylabel("Number of Flights")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# Flight status distribution
plt.figure(figsize=(8, 5))
df["Flight_Status"].value_counts().plot(kind="bar")
plt.title("Flight Status Distribution")
plt.xlabel("Flight Status")
plt.ylabel("Number of Flights")
plt.tight_layout()
plt.show()


# Average delay by airline
plt.figure(figsize=(10, 5))
airline_analysis["Average_Delay"].sort_values(
    ascending=False
).plot(kind="bar")
plt.title("Average Delay by Airline")
plt.xlabel("Airline")
plt.ylabel("Average Delay (Minutes)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# Passenger satisfaction by travel class
plt.figure(figsize=(8, 5))
class_analysis["Average_Satisfaction"].sort_values(
    ascending=False
).plot(kind="bar")
plt.title("Average Passenger Satisfaction by Travel Class")
plt.xlabel("Travel Class")
plt.ylabel("Average Satisfaction")
plt.tight_layout()
plt.show()


# Weather vs delay
plt.figure(figsize=(9, 5))
weather_analysis["Average_Delay"].sort_values(
    ascending=False
).plot(kind="bar")
plt.title("Average Delay by Weather Condition")
plt.xlabel("Weather")
plt.ylabel("Average Delay (Minutes)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ============================================================
# 22. CORRELATION ANALYSIS
# ============================================================

print("\n" + "=" * 80)
print("11. CORRELATION ANALYSIS")
print("=" * 80)

numeric_columns = [
    "Passengers",
    "Seat_Capacity",
    "Average_Ticket_Price",
    "Delay_Minutes",
    "Avg_Baggage_Kg",
    "Passenger_Satisfaction",
    "Load_Factor",
    "Estimated_Revenue"
]

correlation_matrix = df[numeric_columns].corr()

print("\nCorrelation Matrix:")
display(correlation_matrix.round(2))

plt.figure(figsize=(10, 7))
plt.imshow(correlation_matrix, aspect="auto")
plt.colorbar()
plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=45,
    ha="right"
)
plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)
plt.title("Correlation Matrix of Numerical Variables")
plt.tight_layout()
plt.show()


# ============================================================
# 23. FINAL KEY FINDINGS / 8 OBSERVATIONS
# ============================================================

print("\n" + "=" * 80)
print("12. FINAL KEY OBSERVATIONS")
print("=" * 80)

total_flights = len(df)
total_passengers = df["Passengers"].sum()
average_delay = df["Delay_Minutes"].mean()
average_satisfaction = df["Passenger_Satisfaction"].mean()

on_time_percentage = (
    (df["Flight_Status"] == "On Time").sum() / total_flights
) * 100

delayed_percentage = (
    (df["Flight_Status"] == "Delayed").sum() / total_flights
) * 100

top_airline = df["Airline"].value_counts().idxmax()

top_destination = df["Destination"].value_counts().idxmax()

best_airline_satisfaction = airline_analysis[
    "Average_Satisfaction"
].idxmax()

worst_weather_condition = weather_analysis[
    "Average_Delay"
].idxmax()

highest_revenue_airline = df.groupby(
    "Airline"
)["Estimated_Revenue"].sum().idxmax()

print(f"""
1. The dataset contains {total_flights} flight records with a total of
   {total_passengers:,} passengers.

2. The overall average flight delay is {average_delay:.2f} minutes,
   indicating the general delay level across all recorded flights.

3. Approximately {on_time_percentage:.2f}% of flights were recorded as
   On Time, while {delayed_percentage:.2f}% were recorded as Delayed.

4. {top_airline} operates the highest number of flights in the dataset,
   making it the most frequently represented airline.

5. {top_destination} is the most frequently occurring destination,
   indicating relatively high activity toward this destination.

6. {best_airline_satisfaction} has the highest average passenger
   satisfaction among the airlines.

7. {worst_weather_condition} is associated with the highest average
   flight delay among the weather conditions in the dataset.

8. {highest_revenue_airline} has the highest estimated revenue based on
   passengers multiplied by average ticket price.
""")

# ============================================================
# 24. FINAL DATASET PREVIEW
# ============================================================

print("\n" + "=" * 80)
print("FINAL ANALYZED DATASET")
print("=" * 80)

print("\nFinal Shape:", df.shape)

print("\nFinal Columns:")
print(df.columns.tolist())

display(df.head(10))

print("\n" + "=" * 80)
print("ANALYSIS COMPLETED SUCCESSFULLY!")
print("=" * 80)
```
