```python
# ================================================================
# DAY 13 - RESTAURANT BRANCH PERFORMANCE EDA
# Complete Exploratory Data Analysis in ONE CELL
# ================================================================

# -----------------------------
# 1. IMPORT LIBRARIES
# -----------------------------
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Try to use seaborn if available
try:
    import seaborn as sns
    SEABORN_AVAILABLE = True
except ImportError:
    SEABORN_AVAILABLE = False

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 150)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")


# -----------------------------
# 2. LOAD DATASET
# -----------------------------
file_path = "Day13_Restaurant_Branch_Performance_Dataset.csv"

df = pd.read_csv(file_path)

print("=" * 80)
print("DAY 13 - RESTAURANT BRANCH PERFORMANCE EDA")
print("=" * 80)

print("\nDataset loaded successfully!")
print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")


# -----------------------------
# 3. FIRST LOOK AT DATA
# -----------------------------
print("\n" + "=" * 80)
print("1. FIRST 5 RECORDS")
print("=" * 80)
display(df.head())

print("\n" + "=" * 80)
print("LAST 5 RECORDS")
print("=" * 80)
display(df.tail())


# -----------------------------
# 4. DATASET INFORMATION
# -----------------------------
print("\n" + "=" * 80)
print("2. DATASET INFORMATION")
print("=" * 80)

print(df.info())


# -----------------------------
# 5. DATA TYPES
# -----------------------------
print("\n" + "=" * 80)
print("3. DATA TYPES")
print("=" * 80)

display(pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str),
    "Non_Null_Count": df.notna().sum().values,
    "Unique_Values": df.nunique().values
}))


# -----------------------------
# 6. MISSING VALUES
# -----------------------------
print("\n" + "=" * 80)
print("4. MISSING VALUE ANALYSIS")
print("=" * 80)

missing = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (df.isnull().sum() / len(df)) * 100
})

missing = missing.sort_values("Missing_Count", ascending=False)

display(missing)

print("\nTotal missing values:", df.isnull().sum().sum())

if df.isnull().sum().sum() == 0:
    print("No missing values found.")
else:
    print("Missing values are mainly present in:")
    display(missing[missing["Missing_Count"] > 0])


# -----------------------------
# 7. DUPLICATE ANALYSIS
# -----------------------------
print("\n" + "=" * 80)
print("5. DUPLICATE RECORD ANALYSIS")
print("=" * 80)

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

if duplicate_count == 0:
    print("No duplicate records found.")
else:
    print("Duplicate records found.")
    display(df[df.duplicated()])


# -----------------------------
# 8. SUMMARY STATISTICS
# -----------------------------
print("\n" + "=" * 80)
print("6. SUMMARY STATISTICS - NUMERICAL VARIABLES")
print("=" * 80)

display(df.describe().T)


# -----------------------------
# 9. CATEGORICAL SUMMARY
# -----------------------------
print("\n" + "=" * 80)
print("7. CATEGORICAL VARIABLE SUMMARY")
print("=" * 80)

categorical_columns = df.select_dtypes(include="object").columns

for col in categorical_columns:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False))


# -----------------------------
# 10. NUMERICAL VARIABLES
# -----------------------------
print("\n" + "=" * 80)
print("8. NUMERICAL VARIABLES")
print("=" * 80)

numeric_columns = df.select_dtypes(include=np.number).columns.tolist()

print(numeric_columns)


# -----------------------------
# 11. KEY BUSINESS METRICS
# -----------------------------
print("\n" + "=" * 80)
print("9. KEY BUSINESS METRICS")
print("=" * 80)

key_metrics = [
    "Customers",
    "Orders",
    "Revenue",
    "Profit",
    "Marketing_Spend",
    "Staff_Count",
    "Avg_Delivery_Min",
    "Customer_Rating"
]

display(df[key_metrics].describe().T)


# -----------------------------
# 12. DISTRIBUTION ANALYSIS
# -----------------------------
print("\n" + "=" * 80)
print("10. DISTRIBUTION ANALYSIS")
print("=" * 80)

for col in key_metrics:
    plt.figure(figsize=(9, 5))
    plt.hist(df[col].dropna(), bins=20, edgecolor="black")
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.grid(axis="y", alpha=0.3)
    plt.show()


# -----------------------------
# 13. BOXPLOT ANALYSIS
# -----------------------------
print("\n" + "=" * 80)
print("11. BOXPLOT ANALYSIS")
print("=" * 80)

for col in key_metrics:
    plt.figure(figsize=(9, 4))
    plt.boxplot(df[col].dropna(), vert=False)
    plt.title(f"Boxplot of {col}")
    plt.xlabel(col)
    plt.grid(axis="x", alpha=0.3)
    plt.show()


# -----------------------------
# 14. BRANCH PERFORMANCE
# -----------------------------
print("\n" + "=" * 80)
print("12. BRANCH PERFORMANCE ANALYSIS")
print("=" * 80)

branch_performance = df.groupby("Branch").agg(
    Total_Customers=("Customers", "sum"),
    Total_Orders=("Orders", "sum"),
    Total_Revenue=("Revenue", "sum"),
    Total_Profit=("Profit", "sum"),
    Avg_Marketing_Spend=("Marketing_Spend", "mean"),
    Avg_Staff_Count=("Staff_Count", "mean"),
    Avg_Delivery_Time=("Avg_Delivery_Min", "mean"),
    Avg_Customer_Rating=("Customer_Rating", "mean")
).sort_values("Total_Revenue", ascending=False)

display(branch_performance)


# Branch Revenue
plt.figure(figsize=(12, 6))
branch_performance["Total_Revenue"].sort_values().plot(kind="barh")
plt.title("Total Revenue by Branch")
plt.xlabel("Revenue")
plt.ylabel("Branch")
plt.grid(axis="x", alpha=0.3)
plt.show()


# Branch Profit
plt.figure(figsize=(12, 6))
branch_performance["Total_Profit"].sort_values().plot(kind="barh")
plt.title("Total Profit by Branch")
plt.xlabel("Profit")
plt.ylabel("Branch")
plt.grid(axis="x", alpha=0.3)
plt.show()


# Branch Customers
plt.figure(figsize=(12, 6))
branch_performance["Total_Customers"].sort_values().plot(kind="barh")
plt.title("Total Customers by Branch")
plt.xlabel("Customers")
plt.ylabel("Branch")
plt.grid(axis="x", alpha=0.3)
plt.show()


# -----------------------------
# 15. REGION PERFORMANCE
# -----------------------------
print("\n" + "=" * 80)
print("13. REGION PERFORMANCE ANALYSIS")
print("=" * 80)

region_performance = df.groupby("Region").agg(
    Total_Customers=("Customers", "sum"),
    Total_Orders=("Orders", "sum"),
    Total_Revenue=("Revenue", "sum"),
    Total_Profit=("Profit", "sum"),
    Avg_Marketing_Spend=("Marketing_Spend", "mean"),
    Avg_Staff_Count=("Staff_Count", "mean"),
    Avg_Delivery_Time=("Avg_Delivery_Min", "mean"),
    Avg_Customer_Rating=("Customer_Rating", "mean")
).sort_values("Total_Revenue", ascending=False)

display(region_performance)


# Region Revenue
plt.figure(figsize=(10, 6))
region_performance["Total_Revenue"].plot(kind="bar")
plt.title("Total Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.show()


# Region Profit
plt.figure(figsize=(10, 6))
region_performance["Total_Profit"].plot(kind="bar")
plt.title("Total Profit by Region")
plt.xlabel("Region")
plt.ylabel("Profit")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.show()


# -----------------------------
# 16. STORE TYPE PERFORMANCE
# -----------------------------
print("\n" + "=" * 80)
print("14. STORE TYPE PERFORMANCE ANALYSIS")
print("=" * 80)

store_performance = df.groupby("Store_Type").agg(
    Total_Customers=("Customers", "sum"),
    Total_Orders=("Orders", "sum"),
    Total_Revenue=("Revenue", "sum"),
    Total_Profit=("Profit", "sum"),
    Avg_Marketing_Spend=("Marketing_Spend", "mean"),
    Avg_Staff_Count=("Staff_Count", "mean"),
    Avg_Delivery_Time=("Avg_Delivery_Min", "mean"),
    Avg_Customer_Rating=("Customer_Rating", "mean")
).sort_values("Total_Revenue", ascending=False)

display(store_performance)


# Store Type Revenue
plt.figure(figsize=(10, 6))
store_performance["Total_Revenue"].plot(kind="bar")
plt.title("Total Revenue by Store Type")
plt.xlabel("Store Type")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.show()


# Store Type Profit
plt.figure(figsize=(10, 6))
store_performance["Total_Profit"].plot(kind="bar")
plt.title("Total Profit by Store Type")
plt.xlabel("Store Type")
plt.ylabel("Profit")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.show()


# -----------------------------
# 17. CUSTOMERS VS REVENUE
# -----------------------------
print("\n" + "=" * 80)
print("15. CUSTOMERS VS REVENUE")
print("=" * 80)

customer_revenue_corr = df["Customers"].corr(df["Revenue"])

print(f"Correlation between Customers and Revenue: {customer_revenue_corr:.3f}")

plt.figure(figsize=(9, 6))
plt.scatter(df["Customers"], df["Revenue"], alpha=0.7)
plt.title("Customers vs Revenue")
plt.xlabel("Customers")
plt.ylabel("Revenue")
plt.grid(alpha=0.3)
plt.show()


# -----------------------------
# 18. MARKETING SPEND VS REVENUE
# -----------------------------
print("\n" + "=" * 80)
print("16. MARKETING SPEND VS REVENUE")
print("=" * 80)

marketing_revenue_corr = df["Marketing_Spend"].corr(df["Revenue"])

print(f"Correlation between Marketing Spend and Revenue: {marketing_revenue_corr:.3f}")

plt.figure(figsize=(9, 6))
plt.scatter(df["Marketing_Spend"], df["Revenue"], alpha=0.7)
plt.title("Marketing Spend vs Revenue")
plt.xlabel("Marketing Spend")
plt.ylabel("Revenue")
plt.grid(alpha=0.3)
plt.show()


# -----------------------------
# 19. STAFF COUNT VS CUSTOMERS
# -----------------------------
print("\n" + "=" * 80)
print("17. STAFF COUNT VS CUSTOMERS")
print("=" * 80)

staff_customer_corr = df["Staff_Count"].corr(df["Customers"])

print(f"Correlation between Staff Count and Customers: {staff_customer_corr:.3f}")

plt.figure(figsize=(9, 6))
plt.scatter(df["Staff_Count"], df["Customers"], alpha=0.7)
plt.title("Staff Count vs Customers")
plt.xlabel("Staff Count")
plt.ylabel("Customers")
plt.grid(alpha=0.3)
plt.show()


# -----------------------------
# 20. DELIVERY TIME VS CUSTOMER RATING
# -----------------------------
print("\n" + "=" * 80)
print("18. DELIVERY TIME VS CUSTOMER RATING")
print("=" * 80)

delivery_rating_corr = df["Avg_Delivery_Min"].corr(df["Customer_Rating"])

print(f"Correlation between Delivery Time and Customer Rating: {delivery_rating_corr:.3f}")

plt.figure(figsize=(9, 6))
plt.scatter(df["Avg_Delivery_Min"], df["Customer_Rating"], alpha=0.7)
plt.title("Delivery Time vs Customer Rating")
plt.xlabel("Average Delivery Time (Minutes)")
plt.ylabel("Customer Rating")
plt.grid(alpha=0.3)
plt.show()


# -----------------------------
# 21. PROFIT ANALYSIS
# -----------------------------
print("\n" + "=" * 80)
print("19. PROFIT ANALYSIS")
print("=" * 80)

profit_summary = df["Profit"].describe()

display(profit_summary.to_frame("Profit"))

positive_profit = (df["Profit"] > 0).sum()
negative_profit = (df["Profit"] < 0).sum()

print("Records with positive profit:", positive_profit)
print("Records with negative profit:", negative_profit)

plt.figure(figsize=(10, 5))
plt.hist(df["Profit"], bins=20, edgecolor="black")
plt.title("Distribution of Profit")
plt.xlabel("Profit")
plt.ylabel("Frequency")
plt.grid(axis="y", alpha=0.3)
plt.show()


# -----------------------------
# 22. REVENUE VS PROFIT
# -----------------------------
print("\n" + "=" * 80)
print("20. REVENUE VS PROFIT")
print("=" * 80)

revenue_profit_corr = df["Revenue"].corr(df["Profit"])

print(f"Correlation between Revenue and Profit: {revenue_profit_corr:.3f}")

plt.figure(figsize=(9, 6))
plt.scatter(df["Revenue"], df["Profit"], alpha=0.7)
plt.title("Revenue vs Profit")
plt.xlabel("Revenue")
plt.ylabel("Profit")
plt.grid(alpha=0.3)
plt.show()


# -----------------------------
# 23. CORRELATION MATRIX
# -----------------------------
print("\n" + "=" * 80)
print("21. CORRELATION MATRIX")
print("=" * 80)

correlation_columns = [
    "Staff_Count",
    "Marketing_Spend",
    "Customers",
    "Orders",
    "Average_Bill",
    "Revenue",
    "Food_Cost",
    "Operating_Cost",
    "Profit",
    "Avg_Delivery_Min",
    "Customer_Rating",
    "Temperature_C",
    "Humidity_Percent"
]

correlation_matrix = df[correlation_columns].corr()

display(correlation_matrix)


# -----------------------------
# 24. CORRELATION HEATMAP
# -----------------------------
print("\n" + "=" * 80)
print("22. CORRELATION HEATMAP")
print("=" * 80)

plt.figure(figsize=(14, 10))

if SEABORN_AVAILABLE:
    sns.heatmap(
        correlation_matrix,
        annot=True,
        fmt=".2f",
        cmap="coolwarm",
        center=0,
        linewidths=0.5
    )
else:
    plt.imshow(correlation_matrix, cmap="coolwarm", aspect="auto")
    plt.colorbar()
    plt.xticks(range(len(correlation_matrix.columns)), correlation_matrix.columns, rotation=90)
    plt.yticks(range(len(correlation_matrix.columns)), correlation_matrix.columns)

plt.title("Correlation Matrix of Numerical Variables")
plt.tight_layout()
plt.show()


# -----------------------------
# 25. FIND STRONG CORRELATIONS
# -----------------------------
print("\n" + "=" * 80)
print("23. STRONG CORRELATION PAIRS")
print("=" * 80)

corr_pairs = correlation_matrix.where(
    np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool)
).stack()

corr_pairs = corr_pairs.reindex(
    corr_pairs.abs().sort_values(ascending=False).index
)

strong_correlations = corr_pairs[abs(corr_pairs) >= 0.50]

if len(strong_correlations) > 0:
    display(strong_correlations.to_frame("Correlation"))
else:
    print("No correlation pair has an absolute correlation of 0.50 or higher.")


# -----------------------------
# 26. TOP AND LOWEST PERFORMING BRANCH
# -----------------------------
print("\n" + "=" * 80)
print("24. TOP AND LOWEST PERFORMING BRANCH")
print("=" * 80)

top_revenue_branch = branch_performance["Total_Revenue"].idxmax()
low_revenue_branch = branch_performance["Total_Revenue"].idxmin()

top_profit_branch = branch_performance["Total_Profit"].idxmax()
low_profit_branch = branch_performance["Total_Profit"].idxmin()

print(f"Highest revenue branch : {top_revenue_branch}")
print(f"Lowest revenue branch  : {low_revenue_branch}")
print(f"Highest profit branch  : {top_profit_branch}")
print(f"Lowest profit branch   : {low_profit_branch}")


# -----------------------------
# 27. TOP REGION AND STORE TYPE
# -----------------------------
print("\n" + "=" * 80)
print("25. TOP REGION AND STORE TYPE")
print("=" * 80)

top_region = region_performance["Total_Revenue"].idxmax()
top_store_type = store_performance["Total_Revenue"].idxmax()

print(f"Highest revenue region     : {top_region}")
print(f"Highest revenue store type : {top_store_type}")


# -----------------------------
# 28. BEST CUSTOMER RATING
# -----------------------------
print("\n" + "=" * 80)
print("26. CUSTOMER RATING ANALYSIS")
print("=" * 80)

best_rating_branch = branch_performance["Avg_Customer_Rating"].idxmax()
lowest_rating_branch = branch_performance["Avg_Customer_Rating"].idxmin()

print(f"Highest rated branch : {best_rating_branch}")
print(
    f"Highest average rating: "
    f"{branch_performance.loc[best_rating_branch, 'Avg_Customer_Rating']:.2f}"
)

print(f"Lowest rated branch  : {lowest_rating_branch}")
print(
    f"Lowest average rating: "
    f"{branch_performance.loc[lowest_rating_branch, 'Avg_Customer_Rating']:.2f}"
)


# -----------------------------
# 29. DELIVERY TIME ANALYSIS
# -----------------------------
print("\n" + "=" * 80)
print("27. DELIVERY TIME ANALYSIS")
print("=" * 80)

fastest_branch = branch_performance["Avg_Delivery_Time"].idxmin()
slowest_branch = branch_performance["Avg_Delivery_Time"].idxmax()

print(f"Fastest branch : {fastest_branch}")
print(
    f"Average delivery time: "
    f"{branch_performance.loc[fastest_branch, 'Avg_Delivery_Time']:.2f} minutes"
)

print(f"Slowest branch : {slowest_branch}")
print(
    f"Average delivery time: "
    f"{branch_performance.loc[slowest_branch, 'Avg_Delivery_Time']:.2f} minutes"
)


# -----------------------------
# 30. PROMOTION ANALYSIS
# -----------------------------
print("\n" + "=" * 80)
print("28. PROMOTION ANALYSIS")
print("=" * 80)

promotion_analysis = df.groupby("Promotion", dropna=False).agg(
    Average_Revenue=("Revenue", "mean"),
    Average_Profit=("Profit", "mean"),
    Average_Customers=("Customers", "mean"),
    Average_Rating=("Customer_Rating", "mean")
).sort_values("Average_Revenue", ascending=False)

display(promotion_analysis)

plt.figure(figsize=(10, 6))
promotion_analysis["Average_Revenue"].plot(kind="bar")
plt.title("Average Revenue by Promotion")
plt.xlabel("Promotion")
plt.ylabel("Average Revenue")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.show()


# -----------------------------
# 31. WEATHER ANALYSIS
# -----------------------------
print("\n" + "=" * 80)
print("29. WEATHER PERFORMANCE ANALYSIS")
print("=" * 80)

weather_analysis = df.groupby("Weather").agg(
    Average_Revenue=("Revenue", "mean"),
    Average_Profit=("Profit", "mean"),
    Average_Customers=("Customers", "mean"),
    Average_Rating=("Customer_Rating", "mean")
).sort_values("Average_Revenue", ascending=False)

display(weather_analysis)


# -----------------------------
# 32. MONTHLY PERFORMANCE
# -----------------------------
print("\n" + "=" * 80)
print("30. MONTHLY PERFORMANCE")
print("=" * 80)

monthly_performance = df.groupby("Month").agg(
    Total_Revenue=("Revenue", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Customers=("Customers", "sum"),
    Average_Rating=("Customer_Rating", "mean")
)

display(monthly_performance)


# -----------------------------
# 33. DAY-WISE PERFORMANCE
# -----------------------------
print("\n" + "=" * 80)
print("31. DAY-WISE PERFORMANCE")
print("=" * 80)

day_performance = df.groupby("Day").agg(
    Average_Revenue=("Revenue", "mean"),
    Average_Profit=("Profit", "mean"),
    Average_Customers=("Customers", "mean"),
    Average_Rating=("Customer_Rating", "mean")
).sort_values("Average_Revenue", ascending=False)

display(day_performance)


# -----------------------------
# 34. OVERALL BUSINESS METRICS
# -----------------------------
print("\n" + "=" * 80)
print("32. OVERALL BUSINESS METRICS")
print("=" * 80)

total_revenue = df["Revenue"].sum()
total_profit = df["Profit"].sum()
total_customers = df["Customers"].sum()
total_orders = df["Orders"].sum()
average_rating = df["Customer_Rating"].mean()
average_delivery = df["Avg_Delivery_Min"].mean()
total_marketing = df["Marketing_Spend"].sum()

profit_margin = (total_profit / total_revenue) * 100

print(f"Total Revenue          : {total_revenue:,.2f}")
print(f"Total Profit           : {total_profit:,.2f}")
print(f"Total Customers        : {total_customers:,.0f}")
print(f"Total Orders           : {total_orders:,.0f}")
print(f"Total Marketing Spend  : {total_marketing:,.2f}")
print(f"Average Customer Rating: {average_rating:.2f}")
print(f"Average Delivery Time  : {average_delivery:.2f} minutes")
print(f"Overall Profit Margin  : {profit_margin:.2f}%")


# -----------------------------
# 35. AUTOMATIC KEY OBSERVATIONS
# -----------------------------
print("\n" + "=" * 80)
print("33. FINAL DATA-DRIVEN OBSERVATIONS")
print("=" * 80)

# Observation 1
highest_revenue_branch_value = branch_performance.loc[
    top_revenue_branch, "Total_Revenue"
]

print(
    f"1. {top_revenue_branch} is the highest revenue-generating branch, "
    f"with total revenue of {highest_revenue_branch_value:,.2f}."
)

# Observation 2
highest_profit_branch_value = branch_performance.loc[
    top_profit_branch, "Total_Profit"
]

print(
    f"2. {top_profit_branch} generates the highest total profit, "
    f"with profit of {highest_profit_branch_value:,.2f}."
)

# Observation 3
print(
    f"3. {top_region} is the strongest region based on total revenue, "
    f"with revenue of {region_performance.loc[top_region, 'Total_Revenue']:,.2f}."
)

# Observation 4
print(
    f"4. {top_store_type} is the highest revenue-generating store type, "
    f"with total revenue of {store_performance.loc[top_store_type, 'Total_Revenue']:,.2f}."
)

# Observation 5
print(
    f"5. The correlation between Customers and Revenue is "
    f"{customer_revenue_corr:.3f}, indicating the strength and direction "
    f"of the relationship between customer volume and revenue."
)

# Observation 6
print(
    f"6. The correlation between Marketing Spend and Revenue is "
    f"{marketing_revenue_corr:.3f}, showing how marketing investment "
    f"is associated with revenue performance."
)

# Observation 7
print(
    f"7. The correlation between Delivery Time and Customer Rating is "
    f"{delivery_rating_corr:.3f}. This helps indicate whether delivery "
    f"speed is associated with customer satisfaction."
)

# Observation 8
print(
    f"8. Across all records, the average customer rating is "
    f"{average_rating:.2f}/5 and the average delivery time is "
    f"{average_delivery:.2f} minutes."
)


# -----------------------------
# 36. FINAL CONCLUSION
# -----------------------------
print("\n" + "=" * 80)
print("34. CONCLUSION")
print("=" * 80)

print(f"""
The Restaurant Branch Performance dataset contains {df.shape[0]} records
and {df.shape[1]} variables covering branch, region, store type, customers,
orders, revenue, profit, marketing expenditure, staff count, delivery time,
customer ratings, weather and promotion information.

The EDA shows differences in performance across branches, regions and
store types. Revenue and profit were compared with customer volume,
marketing expenditure and other operational factors. Distribution analysis
was used to understand the spread of important business variables, while
the correlation matrix helped identify relationships among numerical
variables.

Overall, the analysis provides useful insights into revenue generation,
profitability, customer demand, marketing effectiveness, operational
performance and customer satisfaction. These findings can help restaurant
management identify high-performing branches and areas that may require
operational improvement.
""")

print("=" * 80)
print("EDA COMPLETED SUCCESSFULLY")
print("=" * 80)
```
