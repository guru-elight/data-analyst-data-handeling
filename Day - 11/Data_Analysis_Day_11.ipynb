```python
import pandas as pd
import numpy as np
```

```python
file_path = "Day11_Messy_Company_Employee_Dataset.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

display(df.head())
```

```python
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
display(df.describe(include="all").T)
```

```python
original_rows = df.shape[0]
original_columns = df.shape[1]

original_missing = df.isnull().sum().sum()
original_duplicates = df.duplicated().sum()

print("========== BEFORE CLEANING ==========")
print("Rows:", original_rows)
print("Columns:", original_columns)
print("Missing Values:", original_missing)
print("Duplicate Records:", original_duplicates)
```

```python
missing_report = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

display(missing_report)
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)

if duplicate_count > 0:
    display(df[df.duplicated(keep=False)].sort_values("Employee_ID"))
```

```python
categorical_columns = [
    "Department",
    "Gender",
    "City",
    "Work_Mode"
]

for column in categorical_columns:
    print("\n" + "=" * 50)
    print(column)
    print(df[column].value_counts(dropna=False))
```

```python
text_columns = df.select_dtypes(include="object").columns

for column in text_columns:
    df[column] = df[column].str.strip()

print("Extra spaces removed from text columns.")

```

```python
df["Department"] = df["Department"].replace({
    "ENGINEERING": "Engineering",
    "engineering": "Engineering"
})

print(df["Department"].value_counts(dropna=False))
```

```python
df["Gender"] = df["Gender"].replace({
    "female": "Female",
    "MALE": "Male"
})

print(df["Gender"].value_counts(dropna=False))
```

```python
df["City"] = df["City"].replace({
    "delhi": "Delhi",
    "DELHI": "Delhi"
})

print(df["City"].value_counts(dropna=False))
```

```python
df["Work_Mode"] = df["Work_Mode"].replace({
    "REMOTE": "Remote",
    "remote": "Remote"
})

print(df["Work_Mode"].value_counts(dropna=False))
```

```python
duplicates_before = df.duplicated().sum()

df = df.drop_duplicates()

duplicates_after = df.duplicated().sum()

print("Duplicates before cleaning:", duplicates_before)
print("Duplicates after cleaning:", duplicates_after)
print("Rows after removing duplicates:", len(df))
```

```python
print("Missing values after removing duplicates:")

display(
    pd.DataFrame({
        "Missing Values": df.isnull().sum(),
        "Percentage": (
            df.isnull().sum() / len(df) * 100
        ).round(2)
    })
)
```

```python
numeric_columns = [
    "Age",
    "Annual_Salary",
    "Experience_Years",
    "Performance_Score"
]

for column in numeric_columns:
    median_value = df[column].median()
    df[column] = df[column].fillna(median_value)

print("Numeric missing values filled using median.")
```

```python
categorical_columns = [
    "Department",
    "Gender",
    "City",
    "Work_Mode"
]

for column in categorical_columns:
    mode_value = df[column].mode()[0]
    df[column] = df[column].fillna(mode_value)

print("Categorical missing values filled using mode.")
```

```python
df["Joining_Date"] = pd.to_datetime(
    df["Joining_Date"],
    errors="coerce"
)

print(df["Joining_Date"].dtype)
```

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Annual_Salary"] = pd.to_numeric(df["Annual_Salary"], errors="coerce")
df["Experience_Years"] = pd.to_numeric(df["Experience_Years"], errors="coerce")
df["Performance_Score"] = pd.to_numeric(df["Performance_Score"], errors="coerce")

print(df.dtypes)
```

```python
print("========== FINAL MISSING VALUE CHECK ==========")

final_missing = df.isnull().sum()

display(
    pd.DataFrame({
        "Missing Values": final_missing,
        "Percentage": (
            final_missing / len(df) * 100
        ).round(2)
    })
)
```

```python
print("Remaining duplicate records:", df.duplicated().sum())
```

```python
print("========== FINAL DATA TYPES ==========")
print(df.dtypes)
```

```python
for column in ["Department", "Gender", "City", "Work_Mode"]:
    print("\n" + "=" * 50)
    print(column)
    print(df[column].value_counts())
```

```python
display(df.describe().T)
```

```python
final_rows = df.shape[0]
final_columns = df.shape[1]
final_missing = df.isnull().sum().sum()
final_duplicates = df.duplicated().sum()

comparison = pd.DataFrame({
    "Metric": [
        "Rows",
        "Columns",
        "Missing Values",
        "Duplicate Records"
    ],
    "Before Cleaning": [
        original_rows,
        original_columns,
        original_missing,
        original_duplicates
    ],
    "After Cleaning": [
        final_rows,
        final_columns,
        final_missing,
        final_duplicates
    ]
})

display(comparison)
```
