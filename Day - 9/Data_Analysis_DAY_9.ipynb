```python
# Day 9 - Processed E-commerce Dataset

import pandas as pd

# --------------------------------------------------
# 1. Load the datasets
# --------------------------------------------------

orders = pd.read_csv("Day9_Orders.csv")
customers = pd.read_csv("Day9_Customers.csv")
products = pd.read_csv("Day9_Products.csv")

print("Orders Dataset:")
display(orders.head())

print("Customers Dataset:")
display(customers.head())

print("Products Dataset:")
display(products.head())


# --------------------------------------------------
# 2. Check the structure of the datasets
# --------------------------------------------------

print("Orders Shape:", orders.shape)
print("Customers Shape:", customers.shape)
print("Products Shape:", products.shape)

print("\nOrders Columns:")
print(orders.columns.tolist())

print("\nCustomers Columns:")
print(customers.columns.tolist())

print("\nProducts Columns:")
print(products.columns.tolist())


# --------------------------------------------------
# 3. Check missing values
# --------------------------------------------------

print("\nMissing Values in Orders:")
print(orders.isnull().sum())

print("\nMissing Values in Customers:")
print(customers.isnull().sum())

print("\nMissing Values in Products:")
print(products.isnull().sum())


# --------------------------------------------------
# 4. Merge Orders with Customers
# --------------------------------------------------

processed_data = pd.merge(
    orders,
    customers,
    on="Customer_ID",
    how="left"
)

print("\nAfter merging Orders and Customers:")
display(processed_data.head())


# --------------------------------------------------
# 5. Merge with Products
# --------------------------------------------------

processed_data = pd.merge(
    processed_data,
    products,
    on="Product_ID",
    how="left"
)

print("\nAfter merging all datasets:")
display(processed_data.head())


# --------------------------------------------------
# 6. Demonstrate concat()
# --------------------------------------------------

# Create a small DataFrame from the first few rows
sample_1 = processed_data.head(3)
sample_2 = processed_data.tail(3)

combined_sample = pd.concat(
    [sample_1, sample_2],
    ignore_index=True
)

print("\nDemonstration of concat():")
display(combined_sample)


# --------------------------------------------------
# 7. Convert Order Date to DateTime
# --------------------------------------------------

processed_data["Order_Date"] = pd.to_datetime(
    processed_data["Order_Date"],
    errors="coerce"
)

print("\nOrder Date after conversion:")
print(processed_data["Order_Date"].head())


# --------------------------------------------------
# 8. Extract DateTime information
# --------------------------------------------------

processed_data["Order_Year"] = processed_data["Order_Date"].dt.year
processed_data["Order_Month"] = processed_data["Order_Date"].dt.month
processed_data["Order_Day"] = processed_data["Order_Date"].dt.day
processed_data["Order_Day_Name"] = processed_data["Order_Date"].dt.day_name()


# --------------------------------------------------
# 9. Use apply() to create a useful column
# --------------------------------------------------

processed_data["Product_Name_Clean"] = processed_data["Product_Name"].apply(
    lambda x: str(x).strip().title()
)


# --------------------------------------------------
# 10. Calculate Total Sales using apply()
# --------------------------------------------------

processed_data["Calculated_Sales"] = processed_data.apply(
    lambda row: row["Quantity"] * row["Unit_Price"],
    axis=1
)


# --------------------------------------------------
# 11. Create a Sales Category
# --------------------------------------------------

processed_data["Sales_Category"] = processed_data["Calculated_Sales"].apply(
    lambda x: "High" if x >= 5000
    else "Medium" if x >= 2000
    else "Low"
)


# --------------------------------------------------
# 12. Select useful columns
# --------------------------------------------------

final_columns = [
    "Order_ID",
    "Order_Date",
    "Order_Year",
    "Order_Month",
    "Order_Day",
    "Order_Day_Name",
    "Customer_ID",
    "Product_ID",
    "Product_Name_Clean",
    "Quantity",
    "Unit_Price",
    "Calculated_Sales",
    "Sales_Category"
]

# Keep only columns that actually exist
final_columns = [
    column for column in final_columns
    if column in processed_data.columns
]

final_data = processed_data[final_columns].copy()


# --------------------------------------------------
# 13. Display the final processed dataset
# --------------------------------------------------

print("\nFinal Processed Dataset:")
display(final_data.head(10))

print("\nFinal Dataset Shape:")
print(final_data.shape)


# --------------------------------------------------
# 14. Check final data types
# --------------------------------------------------

print("\nFinal Dataset Data Types:")
print(final_data.dtypes)


# --------------------------------------------------
# 15. Check missing values
# --------------------------------------------------

print("\nMissing Values in Final Dataset:")
print(final_data.isnull().sum())


# --------------------------------------------------
# 16. Export the processed dataset
# --------------------------------------------------

final_data.to_csv(
    "Processed_Ecommerce_Dataset.csv",
    index=False
)

print("\nProcessed dataset exported successfully.")


# --------------------------------------------------
# 17. Read the exported file to verify
# --------------------------------------------------

verification = pd.read_csv("Processed_Ecommerce_Dataset.csv")

print("\nExported Dataset:")
display(verification.head())

print("\nExported Dataset Shape:")
print(verification.shape)
```
